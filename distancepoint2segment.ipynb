```python
import math
import numpy as np

def point_to_segment_distance(point, segment):
    # """
    # Calculate the shortest distance between a point and a segment.

    # Args:
    #    point (numpy): A numpy representing the coordinates of the point.
    #    segment_start (numpy): A numpy representing the coordinates of the start point of the segment.
    #    segment_end (numpy): A numpy representing the coordinates of the end point of the segment.

    # Returns:
    #    float: The shortest distance between the point and the segment.

    # """
    if len(point) != len(segment[0]) or len(point) != len(segment[1]):
        raise ValueError("Point and segment must have the same number of dimensions")

    AP = point[:] - segment[0][:] 
    AB = segment[1][:] - segment[0][:]

    if np.dot(AB, AB) == 0:
        return np.linalg.norm(AP)

    r = np.dot(AP, AB) / np.dot(AB, AB)

    print(r)
    if r < 0:
        distance_line = np.linalg.norm(AP)
    elif r > 1:
        distance_line = np.linalg.norm(point - segment[1][:])
    else:
        distance_line = np.linalg.norm(point - (segment[0][:] + r * AB))
    

    return distance_line
    
```

```python
point = np.array([1, 3])
segment = np.array([[0, 0], [2, 0]])

distance = point_to_segment_distance(point, segment)

print(distance)
```
